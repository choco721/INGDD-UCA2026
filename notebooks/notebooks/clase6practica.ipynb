## 1. Anatomía de un DAG

Un DAG (grafo acíclico dirigido) declara tareas y el orden entre ellas. Escribimos el DAG a la carpeta `dags/`, que el scheduler de Airflow lee automáticamente.

## 2. Construir el DAG del proyecto

Completá la definición de tareas y sus dependencias. El DAG de referencia completo está en `dags/pipeline_bcra.py`.

````python

dag_code = r'''
from __future__ import annotations

import pendulum
from airflow.models.dag import DAG
from airflow.operators.bash import BashOperator

default_args = {
    "owner": "grupo03",
    "retries": 0,
}

with DAG(
    dag_id="pipeline_grupo03",
    start_date=pendulum.datetime(
        2024, 1, 1,
        tz="America/Argentina/Cordoba",
    ),
    schedule=None,
    catchup=False,
    max_active_runs=1,
    default_args=default_args,
    tags=["proyecto-integrador", "vatsim", "grupo03"],
) as dag:

    extraer = BashOperator(
        task_id="extraer",
        bash_command=r"""
echo "=== DIAGNOSTICO DBT Y MINIO GRUPO03 ==="

echo ""
echo "1. Comprobando dbt:"
if command -v dbt >/dev/null 2>&1; then
    dbt --version
else
    echo "DBT NO ENCONTRADO"
fi

echo ""
echo "2. Comprobando librerias de Python:"
python - <<'PY'
import importlib.util

for paquete in ["duckdb", "dbt", "deltalake", "boto3"]:
    estado = (
        "DISPONIBLE"
        if importlib.util.find_spec(paquete)
        else "NO DISPONIBLE"
    )
    print(f"{paquete}: {estado}")
PY

echo ""
echo "3. Variables de entorno:"
for variable in \
    DBT_DUCKDB_PATH \
    MINIO_BUCKET \
    MINIO_ENDPOINT \
    MINIO_ACCESS_KEY \
    MINIO_SECRET_KEY \
    AWS_ACCESS_KEY_ID \
    AWS_SECRET_ACCESS_KEY \
    AWS_ENDPOINT_URL
do
    if [ -n "${!variable:-}" ]; then
        echo "$variable: CONFIGURADA"
    else
        echo "$variable: NO CONFIGURADA"
    fi
done

echo ""
echo "4. Proyecto dbt:"
ls -ld /opt/airflow/dags/grupo03/dbt
ls -l /opt/airflow/dags/grupo03/dbt/dbt_project.yml

echo ""
echo "DIAGNOSTICO FINALIZADO"
""",
    )
'''

print(dag_code)


````

## 3. Validar la sintaxis del DAG

Antes de desplegarlo, comprobamos que el código Python del DAG compila. Airflow no cargará un DAG con errores de sintaxis.

````python
import ast

# Validar la sintaxis del código del DAG.
ast.parse(dag_code)

print("Sintaxis del DAG válida: no se encontraron errores.")
````

> **Backfill y reintentos** (contenido de la semana 11): con `catchup=True` Airflow ejecuta las corridas históricas pendientes; `retries` reintenta las tareas que fallan.

> **Grupo 03.** Esta celda se usó para desplegar la primera versión del DAG. Hoy el DAG se mantiene directamente en `dags/pipeline_grupo03.py`, así que la celda siguiente guarda el borrador en `respaldos_dags/` como `.txt` y **no toca** el DAG desplegado.

````python
import ast
from pathlib import Path
from entorno import DAGS_DIR

# Validar la sintaxis del borrador antes de guardarlo.
ast.parse(dag_code)

# El DAG real vive en dags/pipeline_grupo03.py y no se sobrescribe desde acá.
# El borrador se guarda como .txt fuera de dags/ para que Airflow no lo lea
# como un segundo DAG con el mismo dag_id.
ruta = Path(DAGS_DIR).parent / "respaldos_dags" / "borrador_clase6.py.txt"
ruta.parent.mkdir(parents=True, exist_ok=True)
ruta.write_text(dag_code, encoding="utf-8")

print("Borrador guardado en:", ruta)
print("DAG desplegado sin cambios:", Path(DAGS_DIR) / "pipeline_grupo03.py")
````

### Versión final desplegada

El DAG que corre en Airflow es `dags/pipeline_grupo03.py`, que evolucionó respecto del borrador de esta clase. La celda anterior guarda el borrador aparte, así que el notebook se puede volver a ejecutar completo sin afectar el DAG desplegado.

Versión final:

- 5 tareas: `preparar_entorno → ingerir_bronze → dbt_run → dbt_test → publicar_gold`.
